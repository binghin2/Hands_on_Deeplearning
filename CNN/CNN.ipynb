```python
import torch
import torch.nn as nn
```

```python
inputs = torch.Tensor(1,1, 28, 28) # Batch size , channel, height, weight
print('텐서의 크기 : {}'.format(inputs.shape))
```

```python
conv1 = nn.Conv2d(1, 32, 3, padding = 1) # input channel, output channel, kernel size = 3x3 
print(conv1)
```

```python
conv2 = nn.Conv2d(32, 64, 3, padding = 1)
print(conv2)
```

```python
pool = nn.MaxPool2d(2)
print(pool)
```

```python
out = conv1(inputs)
print(out.shape)
```

```python
out = pool(out)
print(out.shape)
```

```python
out = conv2(out)
print(out.shape)
```

```python
out = pool(out)
print(out.shape)
```

```python
out = out.view(out.size(0), -1) # out tensor를 펼치는 작업
print(out.shape)
```

```python
fc = nn.Linear(3136, 10) # input dimension = 3136, output dimension = 10
out = fc(out)
print(out.shape)
```
