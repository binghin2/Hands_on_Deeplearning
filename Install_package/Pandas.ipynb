```python
import pandas as pd # 국룰
pd.__version__
```

Pandas는 데이터 처리를 하는 라이브러리임

### 1. Series

```python
sr = pd.Series([17000, 18000, 1000, 5000],
            index = ["피자", "치킨", "콜라", "맥주"])
print('시리즈 출력')
print('-'*15)
print(sr)
# Index와 Value 를 출력가능하다!!
```

```python
print('시리즈의 값 : {}'. format(sr.values))
print('시리즈의 인덱스 : {}'. format(sr.index))
```

### 2. DataFrame

```python
# 2차원 List를 매개변수로 전달함 행이랑 열 존재!! 
values = [[1,2,3], [4,5,6], [7,8,9]] # 1행 123, 2행 456, 3행 789
index =  ['one', 'two', 'three']
columns = ['A', 'B', 'C'] # 1열 A, 2열 B, 3열 C

df = pd.DataFrame(values, index, columns)

print('데이터프레임 출력 ')
print('-'*15)
print(df)
```

```python
# 또한 값 따로 출력가능
print('DataFrame Index: {}'.format(df.index))
print('DataFrame Columns: {}'.format(df.columns))
print('DataFrame Values : ')
print(df.values)
```
