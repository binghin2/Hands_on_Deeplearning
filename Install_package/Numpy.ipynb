```python
import numpy as np # 국룰
np.__version__
```

```python
# 1-dimension array
vector = np.array([1,2,3,4,5])
print(vector)
```

```python
# 2-dimension array
mat = np.array([[10,20,30], [40,50,60]])
print(mat)
```

```python
print("vector 타입 :", type(vector))
print("mat 타입 : ", type(mat))
```

```python
print(vector)
print("vector 의 축 개수 : ", vector.ndim)
print("vector 의 크기(shape)", vector.shape)
```

```python
print(mat)
print("mat 의 축 개수 :", mat.ndim)
print("mat 의 크기(shape) :", mat.shape) #2행3열
```

```python
# 2행3열 영행렬 np.zeros(())
zeromat = np.zeros((2,3))
print(zeromat)
```

```python
# 2행3열 1행렬 np.ones(())
onemat = np.ones((2,3))
print(onemat)
```

```python
# 모두 원하는 값 넣고싶어 np.full((), n)
imat = np.full((2,3), 10)
print(imat)
```

```python
# 대각선이 1이고, 나머지 0인 단위행렬 np.eye(n) => nxn matrix
eyemat = np.eye(3)
print(eyemat)
```

```python
# 정말랜덤값 np.random.random(())
randommat = np.random.random((3,2))
print(randommat)
```

```python
# Range를 가지고 만드는 배열! np.arange()
rangevec = np.arange(10)
print(rangevec)
```

```python
# np.arange(i,j,k) 를 하면 i부터 j-1 까지 k 씩 증가
range_stepvec = np.arange(2,20,4)
print(range_stepvec)
```

```python
# 배열 변경 함수
# .reshape(n,m) 을 하면 n행 m열 matrix로 바꿔준다!!
vec = np.array(np.arange(20))
print(vec)
vec_reshape = vec.reshape(4,5)
print(vec_reshape)
```

```python
mat = np.array(([1,2,3], [4,5,6]))
print(mat)
```

```python
# Slicing 원하는거만 짤라서 볼 수 있다!! mat[0,0]
# e.g. 원하는 행을 보고싶다 mat[n,:] => n+1 행을 출력함
# e.g. 원하는 열을 보고싶다 mat[:,n] => n+1 열을 출력함
smat = mat[0,:] # 1행 출력
print(smat)
scmat = mat[:,0] # 1열 출력
print(scmat)
```

```python
# indexing 원하는 원소만 뽑아와서 배열로 재구성 할 수 있다!!
mat = np.array([[1,2], [4,5], [6,7]])
print(mat)
```

```python
# 2행1열 원소
print(mat[1,0])
```

```python
# 2행1열이랑 3행1열 가지고 만들고싶어
indexing_mat = mat[[1,2], [0,0]]
print(indexing_mat)
```

```python
# 연산 또한 자유롭게 가능 
x = np.array([[1,2], [3,4]])
y = np.array([[2,3], [4,5]])

print(x)
print(y)
print(x*y)
z = np.dot(x,y)
print(z)
# x*y와 np.dot(x,y)가 다른 이유
# 행렬곱임!!
```
