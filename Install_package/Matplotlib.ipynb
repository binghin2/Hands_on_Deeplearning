```python
import matplotlib.pyplot as plt
```

```python
# Line plot
plt.title('test')
plt.plot([1,2,3,4], [2,3,4,5]) # x = 1,2,3,4 y = 2,3,4,5
plt.show()
```

```python
# Label 삽입
plt.title('test')
plt.plot([1,2,3,4], [1,8,2,5])
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
plt.title('2line')
plt.plot([1,2,3,4], [1,2,3,4])
plt.plot([2,3,4,5,6,7], [6,3,1,2,3,5])
plt.xlabel('x')
plt.ylabel('y')
plt.legend(['A line', 'B line']) # 범례 삽입
plt.show()
```
