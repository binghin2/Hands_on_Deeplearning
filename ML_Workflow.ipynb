# Machine Learning Workflow

### 1.데이터 수집
Machine 에 학습을 시켜줘야 할 데이터를 모으는 과정임
NLP의 경우 데이터를 말뭉치 , Corpus라고 함 -> Text 집합임 파일 형식은 txt, csv, xml 등 다양함

### 2. 점검&탐색 (Inspection and Exploration)
데이터 구조 , 노이즈 데이터, 정제 방식에 대해 파악하는 과정
탐색적 데이터 분석(Exploratory Data Analysis, EDA) 라고도 함. 독립변수, 종속변수, 변수 유형, 데이터 타입 점검하는 과정 

### 3. 전처리&정제 (Preprocessing and Cleaning)

### 4. 모델링&훈련 (Modeling and Training)
전처리 후 코드작성하는 단계 
적절한 머신러닝 알고리즘 사용해서 모델링을 하면, 
전처리 된 데이터를 알고리즘을 통해 Training 함. 
주의해야 할 점으로는 모든 데이터를 학습시켜서는 안됨. 일부는 테스트용으로 남겨두고 훈련용 데이터만 사용해야 함
가장 이상적인 것으로는
훈련(Training), 검증(Validation), 테스트(Test) 
이렇게 3가지로 데이터를 나누고 사용하는것이 좋음!

Validation 과 Test 의 차이점
Validation 이란 모의고사, Test 는 수능이라고 생각하면 이해가 쉽다.
검증용 데이터로 학습이 잘 되었는지 판단을 하는 근거로 사용한다. 
테스트용으로 최종 성능을 평가함. 채점단계에 가까운 데이터임


### 5. 평가(Evaluation)
테스트용 데이터로 모델의 성능을 평가하는 것 
기계가 예측한 데이터가 테스트용 데이터의 정답과 얼마나 일치하는 지!!

### 6. 배포(Deployment)
성공적이면 모델이 배포되는 단계임
하지만 아직 하자가 존재하다면 다시 1. 데이터 수집단계로 돌아감
