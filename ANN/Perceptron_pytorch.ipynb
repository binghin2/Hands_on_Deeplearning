```python
import torch
import torch as nn
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
torch.manual_seed(777) 
if device == 'cuda':
    torch.cuda.manual_seed_all(777)
```

```python
# 1. Single Layeer peceptron 으로 XOR 게이트 구현하기
X = torch.FloatTensor([[0,0],[0,1],[1,0],[1,1]]).to(device) # XOR의 입력 값
Y = torch.FloatTensor([[0], [1], [1], [0]]).to(device) # XOR 의 출력
```

```python
# Using sigmoid function 
linear = torch.nn.Linear(2,1, bias=True).to(device) 
sigmoid = torch.nn.Sigmoid()
model = torch.nn.Sequential(linear, sigmoid).to(device)

```

```python
# 0, 1을 예측하는 Binary Classification 이므로 Cost function으로 Cross Entropy function 사용함 => nn.BCELoss()
# Cost function & Optimizer 설정
criterion = torch.nn.BCELoss().to(device)
optimizer = torch.optim.SGD(model.parameters(), lr=1) 
# optimizer는 경사하강법을 사용함 SGD(Stochastic Gradient Descent) parameters()로 모델의 가중치와 편향을 가져옴
# 여기의 parameters()는 model의 parameters()임
```

```python
for step in range(10001):
    optimizer.zero_grad()
    hypothesis = model(X)
    # hypothesis = model(X)  # 예측 값
    # cost function
    cost = criterion(hypothesis, Y)
    # cost는 예측값에서 실제값 Y를 뺀 값의 오차
    cost.backward()
    # Backpropagation (역전파)
    optimizer.step()
    # weight 갱신

    if step % 100 == 0:
        print(step, cost.item())
```

```python
# 왜 Cost 가 줄지않았냐면 
# Single layer 으로는 XOR 를 해결할 수 없음!!
```

```python
# 예측값 확인
with torch.no_grad(): # 기울기 계산 비활성화!! 성능 평가하는 곳이기 때문임
    hypothesis = model(X) # 예측값 계산
    predicted = (hypothesis > 0.5).float() # 예측값이 0.5를 초과하면 1, 아니면 0
    accuracy = (predicted == Y).float().mean() # 예측값과 실제값Y를 비교하고, 비율을계산
    print('모델 출력값(Hypothesis) : ', hypothesis.detach().cpu().numpy()) 
    print('모델 예측(Predicted) : ', predicted.detach().cpu().numpy())
    print('실제값(Y) : ', Y.cpu().numpy())
    print('정확도(Accuracy) : ', accuracy.item())
```
