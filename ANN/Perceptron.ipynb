```python
# 단층 퍼셉트론 (Single Layer Perceptron)
# AND gate
def And_gate(x1, x2):
    w1 = 0.5
    w2 = 0.5
    b = -0.7
    result = x1*w1 + x2*w2 + b
    if result <= 0:
        return 0
    else:
        return 1
```

```python
And_gate(0,0), And_gate(0,1), And_gate(1,1)
```

```python
# NAND gate
def Nand_gate(x1, x2):
    w1 = -0.5
    w2 = -0.5
    b = 0.7
    result = x1*w1 + x2*w2 + b
    if result <= 0:
        return 0
    else:
        return 1
```

```python
Nand_gate(0,0), Nand_gate(0,1), Nand_gate(1,0), Nand_gate(1,1) # NAND 게이트 결과
```

```python
# OR gate
def Or_gate(x1, x2):
    w1 = 0.6
    w2 = 0.6
    b = -0.5
    result = x1*w1 + x2*w2 + b
    if result <= 0:
        return 0
    else:
        return 1
```

```python
Or_gate(0,0), Or_gate(0,1), Or_gate(1,0), Or_gate(1,1) # OR 게이트 결과
```
