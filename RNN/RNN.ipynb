## RNN 
Recurrent Neural Network은
은닉층에서 activation function를 통해 나온 결과를 출력층 방향에도 보내면서,
다시 은닉층 노드의 다음 계산의 input으로 보내는 특징을 가짐

```python
import torch
```

```python
import torch.nn as nn
import matplotlib.pyplot as plt
from torch import optim
```

```python
input_size = 5 # input 의 크기
hidden_size = 8 # hidden state의 크기
```

```python
# input tensor는 ( batch size x time step x input size) 
inputs = torch.Tensor(1, 10, 5)
# batch size 는 1, 10번의 시점동안 , 5차원의 input vector이 들어가도록 정의
```

```python
cell =  nn.RNN(input_size, hidden_size, batch_first = True) 
# nn.RNN(입력 크기, 은닉 상태 크기 정의, batch_first = True는 input tensor의 첫번째 차원이 batch size임을 명시하는 함수)
```

```python
outputs, _status = cell(inputs)
```

```python
print(outputs.shape) # batchsize는 1, 10번의 시점동안, 8차원의 hidden state가 출력되었다는 의미
```

```python
print(_status.shape) # 마지막 시점의 hidden state는 1,1,8 임 batchsize는 1, 1번의 시점, 8차원
```
